# Day 7 – Dataset Summary Report

## Objective
Explore the basic structure and characteristics of the provided student dataset using **Python and Pandas**.

This notebook demonstrates:
- Loading a CSV file
- Inspecting the first and last records
- Viewing random samples
- Checking shape, columns, index, and data types
- Using `info()` and `describe()`
- Checking missing values
- Writing basic observations about the dataset


```python
import pandas as pd
import numpy as np

# Load the CSV dataset
df = pd.read_csv("Day7_Student_Dataset.csv")

# Display the DataFrame
df.head()
```

## 1. First Few Rows


```python
# Display the first 5 rows
df.head()
```

## 2. Last Few Rows


```python
# Display the last 5 rows
df.tail()
```

## 3. Random Sample


```python
# Display 5 random records
df.sample(5, random_state=42)
```

## 4. Dataset Shape


```python
# Number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Shape:", df.shape)
```

## 5. Column Names


```python
# Display all column names
print("Columns:")
for column in df.columns:
    print("-", column)
```

## 6. Dataset Index


```python
# Display the DataFrame index
print(df.index)
```

## 7. Data Types


```python
# Display data type of every column
df.dtypes
```

## 8. Dataset Information using info()


```python
# Detailed information about the dataset
df.info()
```

## 9. Descriptive Statistics


```python
# Descriptive statistics for numerical columns
df.describe()
```

## 10. Descriptive Statistics for All Columns


```python
# Include categorical/object columns as well
df.describe(include="all")
```

## 11. Missing Value Analysis


```python
# Count missing values in each column
missing_values = df.isnull().sum()
missing_values
```

```python
# Display only columns containing missing values
missing_values[missing_values > 0]
```

## 12. Basic Dataset Observations


```python
print("===== DATASET SUMMARY =====")
print(f"Total rows: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")

print("\nFeatures:")
print(", ".join(df.columns))

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 13. Observations

Based on the dataset exploration:

- The dataset contains a fixed number of student records represented by the rows.
- Each column represents a feature/attribute of the students.
- `head()` and `tail()` help inspect the beginning and end of the dataset.
- `sample()` provides random records for quick inspection.
- `shape`, `columns`, and `index` describe the DataFrame structure.
- `dtypes` and `info()` show the data types and non-null information for each feature.
- `describe()` provides statistical summaries such as count, mean, standard deviation, minimum, maximum, and quartiles for numerical features.
- The missing-value analysis identifies whether any feature contains incomplete records.
- Duplicate rows are also checked as part of the basic dataset inspection.


```python
# Automatically generate a concise observation
print(f"The dataset has {df.shape[0]} rows and {df.shape[1]} columns.")

missing_total = int(df.isnull().sum().sum())
if missing_total == 0:
    print("There are no missing values in the dataset.")
else:
    print(f"The dataset contains {missing_total} missing values.")
    print("Columns with missing values:")
    print(df.isnull().sum()[df.isnull().sum() > 0])

print(f"The dataset contains {df.duplicated().sum()} duplicate rows.")
```
